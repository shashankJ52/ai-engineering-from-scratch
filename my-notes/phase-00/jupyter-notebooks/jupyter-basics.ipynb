```python
import pandas as pd

df = pd.DataFrame({
    "model": ["Linear Model", "Random Forest", "Neural Net"],
    "accuracy": [0.72, 0.89, 0.94],
    "training_time_seconds": [0.1, 2.3, 45.6]
})

df
```

```python
plt.figure(figsize=(8, 4))
plt.bar(df["model"], df["accuracy"])
plt.title("Model Accuracy Comparison")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.show()
```
